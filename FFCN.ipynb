```python
import numpy as np

# Step 1: Define input token embedding
d_model = 8   # input size (e.g., from self-attention output)
d_ff = 32     # expanded feedforward size

# Example token vector (e.g., for a single token after attention)
x = np.random.rand(d_model)
print("Input x:\n", x)

# Step 2: Initialize weights and biases for FFCN
W1 = np.random.rand(d_model, d_ff)
b1 = np.random.rand(d_ff)

W2 = np.random.rand(d_ff, d_model)
b2 = np.random.rand(d_model)

# Step 3: First linear transformation (expand dimensions)
z1 = np.dot(x, W1) + b1
print("\nAfter Linear 1 (xW1 + b1):\n", z1)

# Step 4: Apply activation (ReLU)
a1 = np.maximum(0, z1)
print("\nAfter ReLU:\n", a1)

# Step 5: Second linear transformation (project back to d_model)
z2 = np.dot(a1, W2) + b2
print("\nFinal Output of FFCN (a1W2 + b2):\n", z2)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Generate input values
x = np.linspace(-5, 5, 200)  # 200 values between -5 and 5

# Step 2: Define outputs
linear_output = 2 * x + 1                        # Pure linear transformation
relu_output = 2 * np.maximum(0, x) + 1           # Linear after ReLU

# Step 3: Plot both
plt.figure(figsize=(10, 6))

plt.plot(x, linear_output, '--', label='Linear: y = 2x + 1', color='blue')
plt.plot(x, relu_output, '-', label='Non-Linear with ReLU: y = 2·ReLU(x) + 1', color='red')

# Axes and decorations
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.grid(True)
plt.title("ReLU Introduces Non-Linearity")
plt.xlabel("x")
plt.ylabel("Output")
plt.legend()
plt.show()
```
